```python
# Step 2.78: Regression Error by Target Range

range_error_results = []

for target, (actual, predicted) in regression_predictions.items():

    actual_values = actual.values

    q33 = np.percentile(actual_values, 33)
    q67 = np.percentile(actual_values, 67)

    ranges = np.where(
        actual_values <= q33,
        "Low",
        np.where(
            actual_values <= q67,
            "Medium",
            "High"
        )
    )

    temp = pd.DataFrame({
        "Actual": actual_values,
        "Predicted": predicted,
        "Range": ranges
    })

    for value_range in ["Low", "Medium", "High"]:

        subset = temp[temp["Range"] == value_range]

        range_error_results.append({
            "Target": target,
            "Range": value_range,
            "Count": len(subset),
            "MAE": mean_absolute_error(
                subset["Actual"],
                subset["Predicted"]
            ),
            "RMSE": np.sqrt(
                mean_squared_error(
                    subset["Actual"],
                    subset["Predicted"]
                )
            )
        })

range_error_df = pd.DataFrame(range_error_results)

print("Regression Error by Target Range")
print("--------------------------------")
print(range_error_df.round(4))
```
